```python
#student performance analyser

#1. read the csv
#create average column using math science and english
#result column, is avg is greater than 55 it will be pass otherwise fail
#print class average
#average ka max, average ka min
#Print top student or students.
#Print students with attendance below 75.
#Print students who submitted fewer than 6 assignments
```

```python
import pandas as pd
import numpy as np


def get_result(avg):
  if avg >=55:
    return "Pass"
  return "fail"


df= pd.read_csv("session4_dummy_students_data.csv")

df["average"] = df[["math", "science", "english"]].mean(axis=1)
df["result"] = df["average"].apply(get_result)


print("Full Data:")
print (df)


print("Class average:", df["average"].mean())
print("Max average:", df["average"].max())
print("Min average:", df["average"].min())

```
